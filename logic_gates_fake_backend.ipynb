## AND

```python
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit_ibm_runtime.fake_provider import FakeManilaV2
import numpy as np
from itertools import product


def and_gate(circ: QuantumCircuit, q_reg: QuantumRegister, a_reg: QuantumRegister) -> None:

    circ.ccx(q_reg[0], q_reg[1], a_reg[0])

backend = FakeManilaV2()


for a, b in product([0, 1], repeat=2):

    q_reg = QuantumRegister(2, "q")
    a_reg = QuantumRegister(1, "ancilla")
    cl_reg = ClassicalRegister(1, "output")
    circuit = QuantumCircuit(q_reg, a_reg, cl_reg)

    vector = np.zeros(4, dtype=complex)
    index = 2 * b + a
    vector[index] = 1.0
    circuit.initialize(vector, q_reg)

    and_gate(circuit, q_reg, a_reg)

    circuit.measure(a_reg[0], cl_reg[0])

    compiled = transpile(circuit, backend)
    result = backend.run(compiled, shots=2024).result().get_counts()

    output = max(result, key=result.get) 

    expected = a & b
    correct_counts = result.get(str(expected), 0)
    wrong_counts = 2024 - correct_counts
    error_rate = wrong_counts / 2014
    
    print(f"Input ({a},{b}) : Output {output} : Doğru {correct_counts} : Yanlış {wrong_counts} : Hata Oranı %{error_rate * 100:.2f}")
```

## OR

```python
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit_ibm_runtime.fake_provider import FakeManilaV2
import numpy as np
from itertools import product

def or_gate(circ: QuantumCircuit, q_reg: QuantumRegister, a_reg: QuantumRegister) -> None:
    circ.x(q_reg[0])
    circ.x(q_reg[1])
    circ.ccx(q_reg[0], q_reg[1], a_reg[0])
    circ.x(a_reg[0])
    circ.x(q_reg[0])
    circ.x(q_reg[1])

backend = FakeManilaV2()

for a, b in product([0, 1], repeat=2):
    q_reg = QuantumRegister(2, "q")
    a_reg = QuantumRegister(1, "ancilla")
    cl_reg = ClassicalRegister(1, "output")
    circuit = QuantumCircuit(q_reg, a_reg, cl_reg)

    vector = np.zeros(4, dtype=complex)
    index = 2 * b + a
    vector[index] = 1.0
    circuit.initialize(vector, q_reg)

    or_gate(circuit, q_reg, a_reg)
    circuit.measure(a_reg[0], cl_reg[0])

    compiled = transpile(circuit, backend)
    result = backend.run(compiled, shots=2014).result().get_counts()
    output = max(result, key=result.get)

    expected = a | b
    correct_counts = result.get(str(expected), 0)
    wrong_counts = 2024 - correct_counts
    error_rate = wrong_counts / 2014
    
    print(f"Input ({a},{b}) : Output {output} : Doğru {correct_counts} : Yanlış {wrong_counts} : Hata Oranı %{error_rate * 100:.2f}")
```
